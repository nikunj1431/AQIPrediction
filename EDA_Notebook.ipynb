Studying the AQI of Chandigarh with time

We plot the AQI with time 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('Chandigarh_23to25.csv')
df['date'] = pd.to_datetime(df['date'])
plt.plot(df['date'], df['aqi_value'])

axis = plt.gca()

axis.xaxis.set_major_locator(mdates.MonthLocator(interval = 3))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.title('Chandigarh AQI by date')
plt.xlabel('Date')
plt.ylabel('AQI')

plt.gcf().autofmt_xdate()

plt.show()

```

```python
We see the AQI tends to spike around november each year. For a better understanding we plot the AQI for just one year - 2024.
```

```python
df_24 = df[df['date'].dt.year == 2024]
plt.plot(df_24['date'], df_24['aqi_value'])

axis = plt.gca()
axis.xaxis.set_major_locator(mdates.MonthLocator(interval =1))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

plt.title('Chandigarh AQI for 2024')
plt.xlabel('Month')
plt.ylabel('AQI')
plt.gcf().autofmt_xdate()

plt.show()
```

We note the following interesting features:

- The AQI spikes in November, which aligns with thhe timings of stubble burning in Punjab.
- There is a spike in the AQI in the winter months: december through february, due to temperature inversion leading to pollutants getting trapped near the surface.
- The AQI stabilizes notably during the autumn and summer months, and crashes as monsoon arrives, and as monsoon ends, it briefly reaches summer levels until stubble burning starts again.

Let us compare Chandigarh's air quality to Delhi's

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('Delhi_23to25.csv')
df['date'] = pd.to_datetime(df['date'])

df_24 = df[df['date'].dt.year == 2024]
plt.plot(df_24['date'], df_24['aqi_value'])

axis = plt.gca()
axis.xaxis.set_major_locator(mdates.MonthLocator(interval =1))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

plt.title('Delhi AQI for 2024')
plt.xlabel('Month')
plt.ylabel('AQI')
plt.gcf().autofmt_xdate()

plt.show()
```

As expected, we can see that Delhi's Air quality follows the same pattern as Chandigarh's just with slightly higher highs and lows. The baseline AQI during the summmer months is also higher by about ~50 points.

Let us also compare the AQI data with that of Bengaluru, known for having pleasant weather.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('Bengaluru_23to25.csv')
df['date'] = pd.to_datetime(df['date'])

df_24 = df[df['date'].dt.year == 2024]
plt.plot(df_24['date'], df_24['aqi_value'])

axis = plt.gca()
axis.xaxis.set_major_locator(mdates.MonthLocator(interval =1))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

plt.title('Bengaluru AQI for 2024')
plt.xlabel('Month')
plt.ylabel('AQI')
plt.gcf().autofmt_xdate()

plt.show()
```

This is a rather interesting pattern: similar to Chandigarh and Delhi, Bengaluru experiences the most polluted air in the winters, largely attributable to atmospheric inversion. Amusingly, the quality of the air during the most polluted periods in Bengaluru's atmosphere happen to be comparable to the cleanest periods of Delhi's air.

To understand the nature of pollution in these cities, we compare specific pollutants between Delhi and Bengaluru (Chandigarh's air has the same behaviour as Delhi's air albeit in a milder fashion)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df_blr = pd.read_csv('Bengaluru_15to20.csv')
df_blr['Date'] = pd.to_datetime(df_blr['Date'])

df_blr_19 = df_blr[df_blr['Date'].dt.year == 2019]
plt.plot(df_blr_19['Date'], df_blr_19['CO'], label = 'Bengaluru', linestyle = '-.')


df_del = pd.read_csv('Delhi_15to20.csv')
df_del['Date'] = pd.to_datetime(df_blr['Date'])

df_del_19 = df_del[df_blr['Date'].dt.year == 2019]
plt.plot(df_del_19['Date'], df_del_19['CO'], label = 'Delhi', linestyle = '-')

axis = plt.gca()
axis.xaxis.set_major_locator(mdates.MonthLocator(interval =1))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

plt.title('Bengaluru vs Delhi CO levels (2019)')
plt.xlabel('Month')
plt.ylabel('CO levels')
plt.gcf().autofmt_xdate()
plt.legend()

plt.show()
```

This confirms our suspicions. CO generated from stubble burning and heavy industry practicising incomplete combustion shoots up massively in Delhi, while its levels remainn fairly constant and low across most of the year in Bengaluru, except for the monsoon months.
