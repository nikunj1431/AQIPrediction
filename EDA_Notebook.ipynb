Studying the AQI of Chandigarh with time

We plot the AQI with time 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('Chandigarh_23to25.csv')
df['date'] = pd.to_datetime(df['date'])
plt.plot(df['date'], df['aqi_value'])

axis = plt.gca()

axis.xaxis.set_major_locator(mdates.MonthLocator(interval = 3))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.title('Chandigarh AQI by date')
plt.xlabel('Date')
plt.ylabel('AQI')

plt.gcf().autofmt_xdate()

plt.show()

```

```python
We see the AQI tends to spike around november each year. For a better understanding we plot the AQI for just one year - 2024.
```

```python
df_24 = df[df['date'].dt.year == 2024]
plt.plot(df_24['date'], df_24['aqi_value'])

axis = plt.gca()
axis.xaxis.set_major_locator(mdates.MonthLocator(interval =1))
axis.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

plt.title('Chandigarh AQI for 2024')
plt.xlabel('Month')
plt.ylabel('AQI')
plt.gcf().autofmt_xdate()

plt.show()
```

We note the following interesting features:

- The AQI spikes in November, which aligns with thhe timings of stubble burning in Punjab.
- There is a spike in the AQI in the winter months: december through february, due to temperature inversion leading to pollutants getting trapped near the surface.
- The AQI stabilizes notably during the autumn and summer months, and crashes as monsoon arrives, and as monsoon ends, it briefly reaches summer levels until stubble burning starts again.
